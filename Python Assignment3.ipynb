1. While Loop and Control Statements: Number Guessing Game
Aim: To create a number guessing game using a while loop and the control statements else, break, and continue.

```python

import random

# Generate a random number between 1 and 10
secret_number = random.randint(1, 10)

# Maximum number of attempts
attempts = 3

while attempts > 0:
    guess = int(input("Guess the number (between 1 and 10): "))

    # Check whether the guess is in the valid range
    if guess < 1 or guess > 10:
        print("Your guess is out of range. Please guess a number between 1 and 10.")
        continue

    # Check whether the guess is correct
    if guess == secret_number:
        print("Congratulations! You guessed the correct number.")
        break

    elif guess < secret_number:
        print("Too low. Try again.")

    else:
        print("Too high. Try again.")

    attempts -= 1

else:
    print("Better luck next time!")
    print("The correct number was:", secret_number)
```

2. For Loop: Multiplication Table Generator
Aim: To generate and display the multiplication table of a given number from 1 to 10 using a for loop and the range() function.

```python

# Multiplication Table Generator

num = int(input("Enter the number for which you want the multiplication table: "))

for i in range(1, 11):
    result = num * i
    print(num, "x", i, "=", result)
```

3. Function: BMI Calculator
Aim: To calculate the Body Mass Index (BMI) using a user-defined Python function.

```python

# BMI Calculator

def calculate_bmi(weight, height):
    bmi = weight / (height ** 2)
    return bmi

# Get input from the user
weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))

# Calculate BMI using the function
bmi = calculate_bmi(weight, height)

# Display the result
print("Your BMI is:", format(bmi, ".2f"))
```
